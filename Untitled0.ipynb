```python
import nltk
from nltk.corpus import stopwords
import string

def transform_text(text):
    text = text.lower()
    text = nltk.word_tokenize(text)

    y = []
    for word in text:
        if word.isalnum():  # keep only letters and numbers
            if word not in stopwords.words('english') and word not in string.punctuation:
                y.append(word)

    return " ".join(y)
```

```python
import joblib
from sklearn.naive_bayes import MultinomialNB
from sklearn.feature_extraction.text import TfidfVectorizer

# ... (your previous training code where you trained the model and saved it)

# 5. Function to predict new emails
def predict_spam(new_email):
    """Predicts whether a new email is spam or ham."""
    # Load the saved model and vectorizer
    with open('model.pkl', 'rb') as model_file:
        loaded_model = joblib.load(model_file)
    with open('vectorizer.pkl', 'rb') as vectorizer_file:
        loaded_vectorizer = joblib.load(vectorizer_file)

    # Transform the new email using the loaded vectorizer
    new_email_tfidf = loaded_vectorizer.transform([new_email])

    # Make the prediction using the loaded model
    prediction = loaded_model.predict(new_email_tfidf)[0]

    return "Spam" if prediction == 1 else "Ham"  # Assuming 1 represents spam

# Fit the TfidfVectorizer and model before saving
# (Ensure that 'x_train' and 'y_train' are defined and contain your training data)

# 1. Create a TfidfVectorizer object
vectorizer = TfidfVectorizer()

# 2. Fit the vectorizer to your training data and transform it
x_train_tfidf = vectorizer.fit_transform(x_train)

# 3. Initialize the MultinomialNB model
model = MultinomialNB()

# 4. Fit the model using the TF-IDF features and labels
model.fit(x_train_tfidf, y_train)

# 5. Now you can save the model
joblib.dump(model, open('model.pkl', 'wb'))
joblib.dump(vectorizer, open('vectorizer.pkl', 'wb'))

# Example usage:
new_email = "Congratulations! You have won a free prize!"
result = predict_spam(new_email)
print(result)  # Output: Spam (or Ham)
```
